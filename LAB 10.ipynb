## Lab 10 · Q10.1
## Series Σ (n/(n+1))^(n²)

```python
from sympy import *
n=Symbol('n')
an=sympify(input("Enter the nth term of the given series:"))
exp1=factor(an)**(1/n) # (a_n)^(1/n)
L=limit(exp1,n,oo)
print("The limit L of (a_n)^(1/n) is",L)
if L<1:
    print("The given series converges, since L =",L,"< 1")
elif L>1:
    print("The given series diverges, since L =",L,"> 1")
else:
    print("The test fails at L = 1")
```
