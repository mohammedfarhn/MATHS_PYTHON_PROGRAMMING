## LAB 12
## Verification of multiple products of vectors — scalar and cross product

```python
from sympy.vector import *
from sympy import Symbol,symbols
N = CoordSys3D ('N')
a =3*N.i+5*N.j+4*N.k
b =2*N.i+7*N.j+5*N.k
# To find dot product
adotb=dot(a,b)
print("The dot product of vectors = a . b =",adotb)
# To find cross product
acrossb=cross(a,b)
print("The cross product of vectors = a X b =",acrossb)
# Verification
print("a . (a X b) =",dot(a,acrossb)," and b . (a X b) =",dot(b,acrossb))
lhs=a.magnitude()**2*b.magnitude()**2
rhs=adotb**2+acrossb.magnitude()**2
print("Lagrange identity |a|^2|b|^2 = (a.b)^2 + |a X b|^2 :",lhs,"=",rhs,"->",lhs==rhs)
```
