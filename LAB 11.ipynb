## Lab 11 · Q11.1
## Sum to infinity of 1 + 2²/1! + 3²/2! + 4²/3! + …

```python
from sympy import *
var('n')
a=Lambda(n,(n+1)**2/factorial(n))
print("The nth term of the series:")
pprint(a(n))
sn=Sum(a(n),(n,0,oo))
pprint(sn)
Sn=nsimplify(sn).doit()
print("The sum of the infinite series=",Sn)
print("Numerical value =",N(Sn))
```

## Lab 11 · Q11.2
## Exercise 1: 1 + (1²+2²)/2! + (1²+2²+3²)/3! + …

```python
from sympy import *
var('n')
a=Lambda(n,n*(n+1)*(2*n+1)/(6*factorial(n)))
print("The nth term of the series:")
pprint(a(n))
sn=Sum(a(n),(n,1,oo))
pprint(sn)
Sn=nsimplify(sn).doit()
print("The sum of the infinite series=",Sn)
print("Numerical value =",N(Sn))
```
