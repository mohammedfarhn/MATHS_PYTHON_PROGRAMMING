## Convergent, divergent and oscillatory sequences

Lab 6 · Q6.1
Discuss the convergence of (−1)ⁿ

```python
from sympy import *
n=Symbol('n', integer=True, positive=True)
expr=(-1)**n
L=limit_seq(expr,n) # limit of the sequence
if L is None: # fall back to the ordinary limit
    L=limit(expr,n,oo)
print("Given Expression:",expr)
print("limit=",L)
if abs(L)==oo:
    print("The given sequence diverges to",L)
elif isinstance(L,AccumBounds):
    print("The given sequence oscillates between",L.min,"and",L.max)
else:
    print("The given sequence converges to",L)
```
