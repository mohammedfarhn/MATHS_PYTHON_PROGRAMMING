## LAB 14
## Gradient and Laplacian of a scalar function

## Lab 14 · Q14.1
## Gradient of A = x²y + 2xz − 4

```python
from sympy.vector import *
from sympy import symbols
N=CoordSys3D('N')
A=N.x**2*N.y+2*N.x*N.z-4
gradA=gradient(A)
print(" The Gradient of A=",gradA)
```

## Lab 14 · Q14.2
## Laplacian of A = x²y + 2xz − 4

```python
from sympy.vector import *
from sympy import symbols
N=CoordSys3D('N')
A=N.x**2*N.y+2*N.x*N.z-4
lapA=laplacian(A)
print("Laplacian of A = ",lapA)
```

## Lab 14 · Q14.3
## Exercise a: x³ + y³ + z³ − 3xyz

```python
from sympy.vector import *
from sympy import symbols
N=CoordSys3D('N')
A=N.x**3+N.y**3+N.z**3-3*N.x*N.y*N.z
gradA=gradient(A)
print(" The Gradient of A=",gradA)
lapA=laplacian(A)
print("Laplacian of A = ",lapA)
print("Check: div(grad A) =",divergence(gradA))
```

## Lab 14 · Q14.4
## Exercise b: x²y + y²z + z²y

```python
from sympy.vector import *
from sympy import symbols
N=CoordSys3D('N')
A=N.x**2*N.y+N.y**2*N.z+N.z**2*N.y
gradA=gradient(A)
print(" The Gradient of A=",gradA)
lapA=laplacian(A)
print("Laplacian of A = ",lapA)
print("Check: div(grad A) =",divergence(gradA))
```
