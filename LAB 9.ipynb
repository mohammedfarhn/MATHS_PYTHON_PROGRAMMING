## D'Alembert's ratio test

## Lab 9 · Q9.1
## Series 2!/3 + 3!/3² + 4!/3³ + … + (n+1)!/3ⁿ + …

```python
from sympy import *
n=Symbol('n')
a=Lambda(n,sympify(input("Enter the nth term of the series=")))
print('The given series is',a(n))
ratio=combsimp(a(n+1)/a(n))
print('The ratio is',ratio)
L=limit(ratio,n,oo)
print('The limit is',L)
if L<1:
    print("The given series is convergent by D'Alembert's ratio test")
elif L>1:
    print("The given series is divergent by D'Alembert's ratio test")
else:
    print("D'Alembert's ratio test fails and use Raabe's test to verify the convergence")
    L1=limit(n*((a(n)/a(n+1))-1),n,oo)
    print(f"The limit of the series using Raabe's test is {L1}")
    if L1>1:
        print("The given series is convergent by Raabe's test")
    elif L1<1:
        print("The given series is divergent by Raabe's test")
    else:
        print("Both the tests fail")
```
