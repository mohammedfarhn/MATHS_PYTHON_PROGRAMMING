## Convergence of series by comparison with the p-series

## Lab 8 · Q8.1
## Test the series 1/(1·2) + 1/(3·4) + 1/(5·6) + …

```python
from sympy import *
n=Symbol('n')
an=sympify(input("Enter the nth term of the series an:"))
bn=sympify(input("Enter the nth term of the series bn:"))
L=limit(an/bn,n,oo)
print("lim an/bn =",L)
if L.is_finite and L!=0:
    print("Both the series an and bn converges and diverges together")
    print("Use p-series to find the convergence/divergence of bn")
    p=sympify(input("Enter the value of p="))
    if p>1:
        print("Series bn is convergent")
        print("Hence series an is also convergent")
    else:
        print("Series bn is divergent")
        print("Hence series an is also divergent")
else:
    print("The test fails: choose another comparison series bn")
```
