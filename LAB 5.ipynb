## LAB 5
## Finding the limit of a sequence

## Lab 5 · Q5.1
## Limit of the sequence {1/n}

```python
from sympy import *
n=Symbol('n')
expr=sympify(input("Enter the Expression:"))
L=limit(expr,n,oo)
print("The given sequence is exp=",expr)
print("The limit of the sequence=",L)
```
