## Convergence, divergence and oscillation of series (general method)

## Lab 7 · Q7.1
## Series Σ 1/(n(n+1))

```python
from sympy import *
n=Symbol('n',positive=True,integer=True) # number of terms
k=Symbol('k',positive=True,integer=True) # running index of the sum
ak=1/(k*(k+1))
print("The nth term of the series=")
pprint(ak)
sn=simplify(summation(ak,(k,1,n))) # partial sum Sn
print("The partial sum Sn =",sn)
L=limit_seq(sn,n)
if L is None:
    L=limit(sn,n,oo)
print("The limit of the sequence sn:",L)
if abs(L)==oo:
    print("The given series diverges to",L)
elif isinstance(L,AccumBounds):
    print("The given series oscillates between",L.min,"and",L.max)
else:
    print("The given series converges to",L)
```
