## Lab 3 · Q3.1
## Solve p² + 2xp − 3x² = 0

```python
from sympy import *
x, p, C1, Y = symbols('x p C1 Y')
y=Function('y')
eq1=p**2+2*x*p-3*x**2
sol1=solve(eq1, p) # roots: p = x and p = -3x
print('Roots for p:',sol1)
p=diff(y(x), x) # now p means dy/dx
sol2=1
for i in range(len(sol1)):
    deq = dsolve(Eq(p, sol1[i]), y(x), hint = 'default')
    sol2 = (deq.lhs - deq.rhs) * sol2
print(f"The required solutions is {sol2} = 0")
```

## Lab 3 · Q3.2
## Exercise 1: x³p² + 3xyp + 2y² = 0

```python
from sympy import *
x = symbols('x', positive=True)
ys, p = symbols('ys p') # y is treated as an ordinary symbol while solving for p
roots = solve(x**3*p**2 + 3*x*ys*p + 2*ys**2, p)
print('p =', [simplify(r) for r in roots])
u = sqrt(9 - 8*x)
for s in (1, -1):
    # candidate: ln|y| = 3/(2x) + s*[ -u/(2x) + (2/3) ln((3+u)/(3-u)) ]
    lny = 3/(2*x) + s*(-u/(2*x) + Rational(2,3)*log((3 + u)/(3 - u)))
    p_over_y = (-3 + s*u)/(2*x**2) # required value of (dy/dx)/y
    print('sign', s, ': d(ln y)/dx - p/y =', simplify(diff(lny, x) - p_over_y),
          '| numeric at x=0.5:', N((diff(lny, x) - p_over_y).subs(x, 0.5)))
```

## Lab 3 · Q3.3
## Exercise 2: x²p² + xyp − 6y² = 0

```python
from sympy import *
x, p, ys, C1 = symbols('x p ys C1')
y=Function('y')
eq1=x**2*p**2+x*ys*p-6*ys**2
roots=solve(eq1, p) # p = 2y/x and p = -3y/x
print('Roots for p:',roots)
sol=1
for r in roots:
    deq=dsolve(Eq(diff(y(x),x), r.subs(ys,y(x))), y(x))
    print(deq)
    sol=(deq.lhs-deq.rhs)*sol
print('The required solution is',sol,'= 0')
```
