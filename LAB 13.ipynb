## Vector differentiation and the unit tangent vector

## Lab 13 · Q13.1
## x = sin t, y = cos t, z = t

```python
from sympy.vector import *
import sympy as sp
N=CoordSys3D('N')
t=sp.symbols('t')
x=sp.sin(t)
y=sp.cos(t)
z=t
# To find the position vector r= xi+yj+zk
r=x*N.i+y*N.j+z*N.k
print("The vector r =",r)
#To find the first derivative(tangent vector) of r
r1=sp.diff(r,t)
print("First derivative of r =",r1)
#To find the second derivative of vector r
r2=sp.diff(r1,t)
print("Second derivative of r =",r2)
#To find the unit tangent vector
utv=sp.simplify(r1/sp.sqrt(dot(r1,r1)))
print("Unit tangent vector =",utv)
print("Check |T| =",sp.simplify(sp.sqrt(dot(utv,utv))))
```

## Lab 13 · Q13.2
## x = t²+1, y = 4t−3, z = 2t²−6t; unit tangent at t = 2

```python
from sympy.vector import*
import sympy as sp
N=CoordSys3D('N')
t=sp.symbols('t')
x=t**2+1
y=4*t-3
z=2*t**2-6*t
# To find the position vector r=xi+yj+zk
r=x*N.i+y*N.j+z*N.k
print("The vector r=",r)
#To find the first derivative(tangent vector)of r
r1=sp.diff(r,t)
print("Tangent to r at any point =", r1)
#To find the unit tangent vector
utv=sp.simplify(r1/sp.sqrt(dot(r1,r1)))
print("Unit tangent vector at any point =",utv)
#To find the unit tangent vector at t=2
print("The unit tangent vector at t=2 is=",utv.subs(t,2))
```

## Lab 13 · Q13.4
## Exercise 2: x = 2 log t, y = 4t, z = 2t²+1; at t = 1

```python
from sympy.vector import *
import sympy as sp
N=CoordSys3D('N')
t=sp.symbols('t', positive=True)
x=2*sp.log(t)
y=4*t
z=2*t**2+1
# To find the position vector r= xi+yj+zk
r=x*N.i+y*N.j+z*N.k
print("The vector r =",r)
#To find the first derivative(tangent vector) of r
r1=sp.diff(r,t)
print("First derivative of r =",r1)
#To find the second derivative of vector r
r2=sp.diff(r1,t)
print("Second derivative of r =",r2)
#To find the unit tangent vector
utv=sp.simplify(r1/sp.sqrt(dot(r1,r1)))
print("Unit tangent vector =",utv)
print("Check |T| =",sp.simplify(sp.sqrt(dot(utv,utv))))
#To find the unit tangent vector at t=1
print("The unit tangent vector at t=1 is=",utv.subs(t,1))
# simplify each component (the root is (1+2t^2)^2 under the square root)
comps=[sp.simplify(sp.factor(utv.dot(e))) for e in (N.i,N.j,N.k)]
print("Simplified components of T:",comps)
```
