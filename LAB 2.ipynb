## Lab 2 · Q2.1
## Solve y = −px + p²x⁴

```python
from sympy import *
x=symbols('x')
y=Function('y')
C1=symbols('C1')
p=Function('p')
eq1=y(x)+p(x)*x-x**4*p(x)**2 # y + px - p^2 x^4 = 0
eq2=diff(eq1,x) # differentiate w.r.t. x
eq3=eq2.replace(Derivative(y(x),x),p(x)) # replace dy/dx by p
eq4=factor_list(eq3) # factorise
eq5=eq4[1]
factors=[]
for i in range(len(eq5)):
    factors.append(eq5[i][0])
print('Factors:',factors)
soln=dsolve(factors[0],p(x)) # factor containing dp/dx
print('Solution of the given differential equation is:')
pprint(soln)
# substitute p back into the original equation to get y
ps=soln.rhs
print('y =',simplify(-ps*x+ps**2*x**4))
```

## Lab 2 · Q2.2
## Solve 16x² + 2p²y − p³x = 0

```python
from sympy import *
x, C1, Y = symbols('x C1 Y')
y=Function('y')
p=Function('p')
eq1=16*x**2+2*p(x)**2*y(x)-p(x)**3*x
y1=solve(eq1, y(x)) # solve for y
eq2=factor(diff(y1[0], x) - p(x)) # dy/dx - p = 0
sol1=dsolve(eq2, p(x), hint = 'default') # solve for p
print("p=",sol1)
n=len(sol1)
sol2=eq1.subs(p(x), sol1[n-1].rhs) # last root is p = C1*x
print("The required solution = ",sol2)
# verify the singular solution 2y^3 + 27x^4 = 0 : y = -(3/2)*2^(2/3)*x^(4/3) (x>0)
xp=symbols('xp',positive=True)
ys=-Rational(3,2)*2**Rational(2,3)*xp**Rational(4,3)
ps=diff(ys,xp)
print("Singular check:",simplify(16*xp**2+2*ps**2*ys-ps**3*xp))
```

## Lab 2 · Q2.3
## Exercise a: y = 2px − p²

```python
from sympy import *
p,c=symbols('p c')
x=Function('x')
# p + 2(x - p) dp/dx = 0 -> p dx/dp + 2(x - p) = 0
ode=Eq(p*x(p).diff(p)+2*(x(p)-p),0)
sol=dsolve(ode,x(p))
xs=sol.rhs
ys=simplify(2*p*xs-p**2)
print('x =',xs)
print('y =',ys)
# verification: dy/dp must equal p*dx/dp
print('check:',simplify(diff(ys,p)-p*diff(xs,p)))
```

## Lab 2 · Q2.4
## Exercise b: x² + p²x = yp

```python
from sympy import *
p,c=symbols('p c',positive=True)
x=Function('x')
# 2p dx/dp = x - p^2
ode=Eq(2*p*x(p).diff(p),x(p)-p**2)
sol=dsolve(ode,x(p))
xs=sol.rhs
ys=simplify(xs**2/p+p*xs)
print('x =',xs)
print('y =',ys)
print('check:',simplify(diff(ys,p)-p*diff(xs,p)))
```
