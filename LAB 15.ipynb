## LAB 15
## Divergence and curl of a vector function

## Lab 15 · Q15.1
## A = x²yz i + y²zx j + z²xy k

```python
from sympy.vector import *
N=CoordSys3D('N')
A=(N.x**2*N.y*N.z)*N.i+(N.y**2*N.z*N.x)*N.j+(N.z**2*N.x*N.y)*N.k
divA=divergence(A)
print("Divergence of A = ", divA)
curlA=curl(A)
print("Curl of A = ", curlA)
print("Check: div(curl A) =", divergence(curlA))
```
