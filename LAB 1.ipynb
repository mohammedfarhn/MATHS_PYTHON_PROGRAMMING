### Verification of exactness of a differential equation

## Lab 1 · Q1.1
## Exactness and solution of (2xy + 3y)dx + (x² + 3x)dy = 0

```python
from sympy import *
x,y = symbols('x y',real=True)
M = 2*x*y+3*y
N = x**2+3*x
My = diff(M,y)
Nx = diff(N,x)
print('My=',My)
print('Nx=',Nx)
if My == Nx:
 print("The equation is exact")
else:
 print("The equation is not exact")
y=Function('y')
deq=(2*x*y(x)+3*y(x))+(x**2+3*x)*diff(y(x),x)
ysoln=dsolve(deq,y(x),hint='1st_exact')
pprint(ysoln)
```

## Lab 1 · Q1.2
## Initial value problem dy/dx = −(4x+3y+1)/(3x+2y+1), y(1) = −1

```python
from sympy import *
x,y=symbols('x y',real=True)
M=4*x+3*y+1
N=3*x+2*y+1
if diff(M,y)==diff(N,x):
    print("The equation is exact")
else:
    print("The equation is not exact")
y=Function('y')
deq=(4*x+3*y(x)+1)+(3*x+2*y(x)+1)*diff(y(x),x)
ysoln=dsolve(deq,y(x),ics={y(1):-1},hint='1st_exact')
pprint(ysoln)
```

## Lab 1 · Q1.3
## (x+2y−2)dx + (2x−y+3)dy = 0, y(1) = −1, with graph

```python
from sympy import *
import matplotlib.pyplot as plt
x,y=symbols('x y',real=True)
M=x+2*y-2
My=diff(M,y)
print('My=',My)
N=2*x-y+3
Nx=diff(N,x)
print('Nx=',Nx)
if My==Nx:
    print("The equation is exact")
else:
    print("The equation is not exact")
y=Function('y')
deq=(x+2*y(x)-2)+(2*x-y(x)+3)*diff(y(x),x)
ysoln=dsolve(deq,y(x),ics={y(1):-1},hint='1st_exact')
pprint(ysoln)
plot(ysoln.rhs,(x,-5,5))
```

## Lab 1 · Q1.4
## Exercise 1: (y² − x²)dx + 2xy dy = 0

```python
from sympy import *
x,y=symbols('x y',real=True)
M=y**2-x**2
N=2*x*y
print('My=',diff(M,y)); print('Nx=',diff(N,x))
if diff(M,y)==diff(N,x):
    print("The equation is exact")
else:
    print("The equation is not exact")
y=Function('y')
deq=(y(x)**2-x**2)+2*x*y(x)*diff(y(x),x)
ysoln=dsolve(deq,y(x),hint='1st_exact')
pprint(ysoln)
```

## Lab 1 · Q1.5
## Exercise 2: (x + y³/x²)dx − (3y²/x)dy = 0

```python
from sympy import *
x,y=symbols('x y',real=True)
M=x+y**3/x**2
N=-3*y**2/x
print('My=',simplify(diff(M,y))); print('Nx=',simplify(diff(N,x)))
print("Exact" if simplify(diff(M,y)-diff(N,x))==0 else "Not exact")
# potential function F(x,y): F = integral(M dx) + g(y)
F=integrate(M,x)
g=integrate(simplify(N-diff(F,y)),y)
F=simplify(F+g)
print('F(x,y) =',F,'= C')
# particular solution for y(1)=1 -> C = F(1,1)
C=F.subs({x:1,y:1})
print('C =',C)
```
