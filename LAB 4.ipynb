### Singular solution using Clairaut's form

## Lab 4 · Q4.1
## Solve xp² − yp + p² + 1 = 0

```python
from sympy import *
p,x,y=symbols('p x y',positive=True, real=True)
a,c=symbols('a c')
deq=x*p**2-y*p+p**2+1
c1=deq.subs(p,c) # general solution: replace p by c
print('General solution of given differential equation is:')
print(c1,'=0')
c2=diff(deq,p) # differentiate w.r.t. p (i.e. c)
c3=solve(c2,p) # solve for p
sing_soln=simplify(deq.subs(p,c3[0]))
print('Singular solution of given differential equation is:')
print(sing_soln,'=0')
print('i.e. y =',solve(sing_soln,y))
```

## Lab 4 · Q4.2
## Exercise 1: y + p² = xp + 1

```python
from sympy import *
p,x,y=symbols('p x y',positive=True, real=True)
c=symbols('c')
deq=y-x*p+p**2-1 # y + p^2 - xp - 1 = 0
c1=deq.subs(p,c)
print('General solution:',c1,'=0')
c3=solve(diff(deq,p),p)
sing=simplify(deq.subs(p,c3[0]))
print('Singular solution:',sing,'=0')
```

## Lab 4 · Q4.3
## Exercise 2: y = px + 2p²

```python
from sympy import *
p,x,y=symbols('p x y',real=True)
c=symbols('c')
deq=y-p*x-2*p**2 # y - px - 2p^2 = 0
c1=deq.subs(p,c)
print('General solution:',c1,'=0')
c3=solve(diff(deq,p),p)
sing=simplify(deq.subs(p,c3[0]))
print('Singular solution:',sing,'=0')
```
